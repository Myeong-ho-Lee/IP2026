```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
```

```python
def saturation(value):
  if value > 255:
    value = 255
  elif value < 0:
    value = 0
  return value
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/aero2.bmp')
G_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2_imshow(G_img)
```

```python
gOut = np.zeros((G_img.shape[0], G_img.shape[1]), dtype=np.ubyte)
```

```python
min_pixel_val = 256
max_pixel_val = 0

for h in range(G_img.shape[0]):
  for w in range(G_img.shape[1]):
    if G_img[h, w] < min_pixel_val:
      min_pixel_val = G_img[h, w]
    if G_img[h, w] > max_pixel_val:
      max_pixel_val = G_img[h, w]

print(min_pixel_val, max_pixel_val)
```

```python
for h in range(G_img.shape[0]):
  for w in range(G_img.shape[1]):
    val = (G_img[h, w] - min_pixel_val) / (max_pixel_val - min_pixel_val) * 255
    gOut[h, w] = val

cv2_imshow(gOut)
```
