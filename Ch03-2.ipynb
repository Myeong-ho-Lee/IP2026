```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabIP26/Images/girl.jpg')
G_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2_imshow(G_img)
```

```python
gPlus = np.zeros((G_img.shape[0], G_img.shape[1]), dtype=np.ubyte)
```

```python
val = 140

for h in range(G_img.shape[0]):
  for w in range(G_img.shape[1]):
    plusVal = np.int32(G_img[h][w]) + val
    if(plusVal > 255):
      plusVal = 255
    gPlus[h][w] = plusVal

cv2_imshow(gPlus)
```

```python
gOrigin = np.zeros((G_img.shape[0], G_img.shape[1]), dtype=np.ubyte)
```

```python
val = 140

for h in range(G_img.shape[0]):
  for w in range(G_img.shape[1]):
    plusVal = np.int32(gPlus[h][w]) - val
    if(plusVal > 255):
      plusVal = 255
    gOrigin[h][w] = plusVal

cv2_imshow(gOrigin)
```
